# 단어 유사도 추정에서 PPMI 가중치와 절단 SVD의 효과 분석

## 1. 연구 목적

본 프로젝트에서는 영어 문장에서 단어가 주변 단어와 함께 등장하는
패턴을 바탕으로 단어 벡터를 구성한다. 이를 이용해 PPMI 가중치 적용과
절단 특잇값 분해(Truncated SVD)를 통한 차원 축소가 단어 유사도 추정에
어떤 영향을 미치는지 살펴본다.

## 2. 연구 질문

1. 동시출현 빈도에 PPMI 가중치를 적용하면, 단어 쌍의 유사도 순위가
   사람의 평가에 따른 순위와 더 비슷해지는가?
2. PPMI 행렬에 절단 SVD를 적용하여 단어 벡터의 차원을 줄이면,
   사람의 평가와의 일치도가 어떻게 달라지는가?

## 3. 실험 개요

동일한 영어 말뭉치에서 세 가지 단어 벡터 표현을 비교한다.

1. 동시출현 빈도 기반 벡터
2. PPMI 가중치를 적용한 벡터
3. PPMI 가중치를 적용한 뒤 절단 SVD로 차원을 축소한 벡터

세 방법 모두 동일한 단어 쌍에 대해 코사인 유사도를 계산한다.
이후 스피어만 순위 상관계수를 사용하여 각 방법이 산출한 유사도 순위가
사람의 평가 점수에 따른 순위와 얼마나 일치하는지 비교한다.

```python
   import nltk
from nltk.corpus import brown

# Brown 말뭉치를 다운로드한다.
nltk.download("brown", quiet=True, raise_on_error=True)

# 문장 단위로 단어를 읽고, 알파벳 문자로만 이루어진 단어를 소문자로 변환한다.
sentences = []
for sentence in brown.sents():
    cleaned_sentence = [word.lower() for word in sentence if word.isalpha()]
    if cleaned_sentence:
        sentences.append(cleaned_sentence)

# 전처리 후 문장 수, 전체 토큰 수, 고유 단어 수를 확인한다.
token_count = sum(len(sentence) for sentence in sentences)
vocabulary = {word for sentence in sentences for word in sentence}

print(f"전처리 후 문장 수: {len(sentences):,}")
print(f"전체 토큰 수: {token_count:,}")
print(f"고유 단어 수: {len(vocabulary):,}")
print("첫 번째 문장 예시:", " ".join(sentences[0]))
```

## 4. 단어 동시출현 행렬 구축

전처리된 Brown 말뭉치에서 5회 이상 등장한 단어를 분석 어휘로
선정한다. 각 중심 단어를 기준으로 같은 문장 안에서 왼쪽과 오른쪽
최대 2개 위치에 있는 단어를 문맥으로 사용한다.

동시출현 행렬의 행은 중심 단어를, 열은 문맥 단어를 나타낸다.
각 원소에는 해당 두 단어가 설정된 범위 안에서 함께 등장한 횟수를
기록한다. 이 행렬을 비교 실험의 기준으로 사용한다.

```python
from collections import Counter
from scipy.sparse import csr_matrix

min_frequency = 5
window_size = 2

# 단어별 등장 횟수를 계산하고 분석 어휘에 번호를 부여한다.
word_counts = Counter(word for sentence in sentences for word in sentence)
analysis_words = sorted(
    word for word, count in word_counts.items() if count >= min_frequency
)
word_to_id = {word: index for index, word in enumerate(analysis_words)}

# 같은 문장 안에서 중심 단어와 문맥 단어의 동시출현을 센다.
pair_counts = Counter()

for sentence in sentences:
    word_ids = [word_to_id.get(word) for word in sentence]

    for position, center_id in enumerate(word_ids):
        if center_id is None:
            continue

        start = max(0, position - window_size)
        end = min(len(word_ids), position + window_size + 1)

        for context_position in range(start, end):
            if context_position == position:
                continue

            context_id = word_ids[context_position]
            if context_id is not None:
                pair_counts[center_id, context_id] += 1

# 관측된 단어 쌍의 빈도를 희소 행렬로 저장한다.
rows, columns = zip(*pair_counts)
cooccurrence_matrix = csr_matrix(
    (list(pair_counts.values()), (rows, columns)),
    shape=(len(analysis_words), len(analysis_words)),
)

print(f"분석 어휘 수: {len(analysis_words):,}")
print(f"동시출현 행렬 크기: {cooccurrence_matrix.shape}")
print(f"0이 아닌 원소 수: {cooccurrence_matrix.nnz:,}")
```

## 5. PPMI 가중치 적용

단어의 동시출현 빈도는 각 단어가 얼마나 자주 등장하는지에
영향을 받는다. 이를 고려하기 위해 중심 단어와 문맥 단어가
실제로 함께 등장한 횟수를 두 단어가 독립적으로 등장한다고
가정했을 때의 기대 횟수와 비교한다.

두 횟수의 비율에 밑이 2인 로그를 취한 뒤, 음수는 0으로
처리하여 PPMI(Positive Pointwise Mutual Information)를 구한다.
관측되지 않은 단어 쌍의 가중치는 0으로 둔다.

앞에서 구축한 동시출현 행렬에 이 가중치를 적용하여 두 번째
실험에 사용할 단어 벡터를 만든다.

```python
import numpy as np

# 각 원소의 행과 열 번호를 사용해 관측 빈도와 기대 빈도를 비교한다.
counts = cooccurrence_matrix.tocoo()
row_totals = np.asarray(cooccurrence_matrix.sum(axis=1)).ravel()
column_totals = np.asarray(cooccurrence_matrix.sum(axis=0)).ravel()
total_count = cooccurrence_matrix.sum()

expected_counts = (
    row_totals[counts.row].astype(float) * column_totals[counts.col]
    / total_count
)

# PMI를 계산한 뒤 음수를 0으로 바꾼다.
pmi_values = np.log2(counts.data / expected_counts)
ppmi_values = np.maximum(pmi_values, 0)

# PPMI 행렬을 희소 형식으로 저장하고 값이 0인 항목을 제거한다.
ppmi_matrix = csr_matrix(
    (ppmi_values, (counts.row, counts.col)),
    shape=cooccurrence_matrix.shape,
)
ppmi_matrix.eliminate_zeros()

print(f"PPMI 행렬 크기: {ppmi_matrix.shape}")
print(f"PPMI가 양수인 원소 수: {ppmi_matrix.nnz:,}")
```

## 6. 절단 SVD를 적용한 차원 축소

PPMI 행렬에 절단 특잇값 분해(Truncated SVD)를 적용하여
각 단어를 100차원 벡터로 표현한다. 분석 대상 단어와 행의 순서는
그대로 유지되며, 각 단어를 나타내는 벡터의 차원은 13,366에서
100으로 줄어든다.

이 벡터를 세 번째 실험에 사용한다. 이후 동일한 단어 쌍에 대해
동시출현 빈도, PPMI, PPMI와 절단 SVD를 적용한 표현의
코사인 유사도를 각각 계산한다.

```python
from sklearn.decomposition import TruncatedSVD

# PPMI 행렬을 100차원 단어 벡터로 변환한다.
svd = TruncatedSVD(n_components=100, random_state=42)
svd_embeddings = svd.fit_transform(ppmi_matrix)

print(f"차원 축소 전 행렬 크기: {ppmi_matrix.shape}")
print(f"차원 축소 후 행렬 크기: {svd_embeddings.shape}")
```

## 7. SimLex-999를 이용한 단어 유사도 평가

SimLex-999의 사람 평가 점수를 기준으로 세 가지 단어 벡터 표현의
성능을 비교한다. 두 단어가 모두 분석 어휘에 포함되고, 세 방법에서
두 단어의 벡터가 모두 영벡터가 아닌 단어 쌍만 평가에 사용한다.
세 방법에는 동일한 단어 쌍을 사용하며, 전체 단어 쌍 중 평가에
사용된 비율도 함께 보고한다.

각 단어 쌍의 코사인 유사도를 계산한 뒤, 사람의 평가 점수와
스피어만 순위 상관계수(ρ)를 구한다. ρ가 클수록 해당 방법의
유사도 순위가 사람의 평가 순위에 더 가깝다.

```python
from io import BytesIO
from zipfile import ZipFile

import pandas as pd
import requests
from scipy.sparse import issparse
from scipy.stats import spearmanr
from IPython.display import display

# SimLex-999의 단어 쌍과 사람 평가 점수를 불러온다.
url = "https://fh295.github.io/SimLex-999.zip"
response = requests.get(url, timeout=60)
response.raise_for_status()

with ZipFile(BytesIO(response.content)) as archive:
    data_path = next(
        name for name in archive.namelist()
        if name.rsplit("/", 1)[-1] == "SimLex-999.txt"
    )
    with archive.open(data_path) as file:
        benchmark = pd.read_csv(file, sep="\t")

# 두 단어가 모두 분석 어휘에 포함된 단어 쌍을 선택한다.
evaluation = benchmark[["word1", "word2", "SimLex999"]].copy()
evaluation["id1"] = evaluation["word1"].str.lower().map(word_to_id)
evaluation["id2"] = evaluation["word2"].str.lower().map(word_to_id)
evaluation = evaluation.dropna(subset=["id1", "id2"]).copy()

left_ids = evaluation["id1"].to_numpy(dtype=int)
right_ids = evaluation["id2"].to_numpy(dtype=int)

representations = {
    "동시출현 빈도": cooccurrence_matrix,
    "PPMI": ppmi_matrix,
    "PPMI + 절단 SVD": svd_embeddings,
}

# 각 단어 쌍의 코사인 유사도를 계산한다.
for method, matrix in representations.items():
    left = matrix[left_ids]
    right = matrix[right_ids]

    if issparse(matrix):
        dot_products = np.asarray(left.multiply(right).sum(axis=1)).ravel()
        left_norms = np.sqrt(np.asarray(left.multiply(left).sum(axis=1)).ravel())
        right_norms = np.sqrt(np.asarray(right.multiply(right).sum(axis=1)).ravel())
    else:
        dot_products = (left * right).sum(axis=1)
        left_norms = np.linalg.norm(left, axis=1)
        right_norms = np.linalg.norm(right, axis=1)

    denominators = left_norms * right_norms
    evaluation[method] = np.divide(
        dot_products,
        denominators,
        out=np.full(len(evaluation), np.nan),
        where=denominators > 0,
    )

# 세 방법에서 모두 유사도를 계산할 수 있는 단어 쌍으로 비교한다.
methods = list(representations)
evaluation = evaluation.dropna(subset=methods).copy()

results = []
for method in methods:
    rho = spearmanr(evaluation["SimLex999"], evaluation[method])[0]
    results.append({"표현 방법": method, "스피어만 ρ": rho})

results_table = pd.DataFrame(results)

print(f"전체 단어 쌍 수: {len(benchmark):,}")
print(f"평가에 사용한 단어 쌍 수: {len(evaluation):,}")
print(f"평가 포함 비율: {len(evaluation) / len(benchmark):.1%}")
display(results_table.round(4))
```

## 8. 실험 결과 및 해석

SimLex-999의 전체 999개 단어 쌍 중 세 방법에서 공통으로
유사도를 계산할 수 있는 997개를 평가에 사용하였다.
평가 포함 비율은 99.8%였다.

동시출현 빈도 기반 표현의 스피어만 순위 상관계수는 0.0080이었으며,
PPMI 가중치를 적용한 후에는 0.1192로 증가하였다.
세 방법 중 PPMI가 사람의 유사도 평가 순위와 가장 높은 일치도를
보였으나, 상관계수의 절대적인 크기는 낮았다.

PPMI 행렬에 절단 SVD를 적용하여 벡터 차원을 13,366에서 100으로
줄였을 때, 상관계수는 0.1174였다. 이는 PPMI 단독 결과보다
0.0018 낮은 값으로, 본 실험에서 관찰된 두 방법의 차이는 작았다.

```python
# 사람의 유사도 평가가 높은 쌍과 낮은 쌍을 비교한다.
example_columns = ["word1", "word2", "SimLex999", *methods]

for label, example_pairs in [
    ("사람의 유사도 평가가 높은 단어 쌍", evaluation.nlargest(5, "SimLex999")),
    ("사람의 유사도 평가가 낮은 단어 쌍", evaluation.nsmallest(5, "SimLex999")),
]:
    print(label)
    display(
        example_pairs[example_columns]
        .rename(columns={"word1": "단어 1", "word2": "단어 2", "SimLex999": "사람 평가"})
        .round(4)
        .reset_index(drop=True)
    )
```

## 9. 단어 쌍 사례 분석

사람은 ‘vanish–disappear’에 9.80점, ‘new–ancient’에 0.23점을
부여하였다. 동시출현 빈도 기반 표현에서는 ‘new–ancient’의
코사인 유사도(0.7860)가 ‘vanish–disappear’(0.1961)보다 높았다.
PPMI를 적용한 뒤에는 ‘vanish–disappear’(0.0526)가
‘new–ancient’(0.0273)보다 높아져 사람의 평가 순서와 일치하였다.

한편 ‘creator–maker’와 ‘hole–agreement’의 경우, PPMI에서는
‘creator–maker’(0.0378)가 ‘hole–agreement’(0.0142)보다 높았다.
절단 SVD를 적용한 뒤에는 ‘creator–maker’(0.2808)가
‘hole–agreement’(0.2842)보다 근소하게 낮아져 두 단어 쌍의
순서가 바뀌었다.

전체 997개 단어 쌍에서 PPMI와 PPMI + 절단 SVD의 스피어만
순위 상관계수는 각각 0.1192와 0.1174였다. 전체 상관계수의
차이는 작았지만, 차원 축소 후 개별 단어 쌍의 순위는 달라질 수
있음을 위 사례에서 확인하였다.

## 10. 결론 및 한계

Brown 말뭉치에서 구축한 세 가지 단어 표현을 SimLex-999의
997개 단어 쌍으로 평가하였다. 동시출현 빈도에 PPMI 가중치를
적용하자 스피어만 순위 상관계수가 0.0080에서 0.1192로
높아졌다. PPMI 행렬을 절단 SVD로 100차원으로 축소한 결과,
상관계수는 0.1174였다. 이번 실험에서는 차원을 크게 줄인 뒤에도
사람의 평가 순위와의 일치도가 PPMI 단독 결과와 비슷했다.

다만 세 방법 모두 사람의 평가와 상관이 낮았다. 또한 하나의
말뭉치와 하나의 문맥 범위, SVD 차원 설정을 사용했으므로,
다른 말뭉치나 설정에서도 같은 경향이 나타나는지는 추가
실험을 통해 확인할 필요가 있다.
